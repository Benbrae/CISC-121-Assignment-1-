Python Code of Code 2: PermMissingElem

```python
def solution(A):
  N = len(A) # length of A
  expected_total = 0 # starts a running total at 0
  for i in range(1, N + 2): # Counting loop, starting at 1 until N + 1
    expected_total += i # adds the current count to the total
  actual_total = 0 # starts another running total
  for num in A: # Goes through the list and stores the numbers
    actual_total += num # adds them to the running total
  return expected_total - actual_total # gives back the missing element
```

Test Case 1: Example from the Problem

```python
A = [2, 3, 1, 5]
print(solution(A))
```

Test Case 2: Empty List

```python
A = []
print(solution(A))
```

Test Case 3: Largest Number Missing

```python
A = [1,2,3]
print(solution(A))
```

Test Case 4: Smallest Number Missing

```python
A = [2,3,4]
print(solution(A))
```

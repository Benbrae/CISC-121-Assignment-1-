Python for Code 5: MaxCounters

```python
def solution(N, A):
  counters = [0]*N # makes N counters, all starting at 0
  biggest = 0 # keeps track of the biggest counter so far
  floor = 0 # the lowest any counter is allowed to be
  for X in A: # goes through the steps one at a time
    if X == N + 1: # max counter step
      floor = biggest # raises the floor to the biggest value
    else:
      if counters[X-1] < floor: # checks if the counter is below the floor
        counters[X-1] = floor # raises it to the floor
      counters[X-1] += 1 # adds 1 to the counter
      if counters[X-1] > biggest: # checks if this is now the biggest counter
        biggest = counters[X-1] # if yes, updates the biggest
  for C in range(len(counters)): # goes through each counter by position
    if counters[C] < floor: # checks if the counter is still below the floor
      counters[C] = floor # raises it to the floor
  return counters # gives back the final counters
```

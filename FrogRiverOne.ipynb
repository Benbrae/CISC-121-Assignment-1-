Python for Code 3: FrogRiverOne

```python
def solution(X, A):
  leaf_spots = set() # starts an empty set of where the leafs land
  for K in range(len(A)): # goes through each spot in the list one at a time
    leaf_spots.add(A[K]) # adds the leafs position to the set, repeats are ignored
    if len(leaf_spots) == X: # checks if all positions from 1 to X are covered
      return K #gives back the current second
  return -1 # river was not fully covered so frog cannot cross
```

Test Case 1: The Example

```python
A = [1, 3, 1, 4, 2, 3, 5, 4]
print(solution(5, A))
```

Test Case 2: Not Fully Covered

```python
A = [1, 3, 1, 4, 2]
print(solution(5, A))
```

Test Case 3: Immediate Crossing

```python
A = [1]
print(solution(1, A))
```

Python for Code 4: PassingCars

```python
def solution(A):
  east_direction = 0 # starts a count of the cars heading in the east direction
  pairs = 0 # starts a count of the pairs of passing cars
  for car in A: # goes through the cars from left to right, one at a time
    if car == 0: # checks if the car is heading east
      east_direction += 1 # if yes add to the count
    else:
      pairs += east_direction #a car heading west passes an eastern heading car, add to pairs
      if pairs > 1000000000: # checks if over 1 billion
        return -1 # over 1 bill pairs so -1
  return pairs # returns the number of pairs
```

Test Case 1: The Example

```python
A = [0, 1, 0, 1, 1]
print(solution(A))
```

Test Case 2: No Pairs

```python
A = [1,0]
print(solution(A))
```

Test Case 3: East Then West

```python
A = [0,0,1,1]
print(solution(A))
```

Test Case 4: Over The 1 Billion

```python
A = [0]*50000 + [1]*50000 # combines a list of 50000 zeros and 50000 ones, which is too many
print(solution(A))
```

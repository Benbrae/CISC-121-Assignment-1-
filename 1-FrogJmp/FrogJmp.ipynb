Python Code of Code 1: FrogJmp

```python
def solution(X,Y,D):
  distance = Y - X # the distance the frog has to jump to
  jumps_completed = distance // D # the amount of jumps that fit perfectly in the distance, if remainder it will round down
  remaining_distance = distance % D # the remainder
  if remaining_distance > 0:
    jumps_completed += 1 # adds a jump if the distance has not been fully covered
  return jumps_completed # gives the full number of jumps completed to cross the distance
```

Test Case 1: Evenly Divided

```python
X = 10
Y = 70
D = 30

print(solution(X,Y,D))
```

Test Case 2: The Frog is Already There

```python
X = 20
Y = 20
D = 30

print(solution(X,Y,D))
```

Test Case 3: Huge Distance and Tiny Jumps

```python
X = 1
Y = 1000000000
D = 1

print(solution(X,Y,D))
```

```python
X = 2
Y = 8.5
D = 74

print(solution(X,Y,D))
```
